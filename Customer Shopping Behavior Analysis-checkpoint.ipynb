```python
#here we found that Review Rating is having 37 missing values in this data set
```

```python
# Imputed missing values in the Review Rating column using the median rating of each product category.
```

```python
data['Review Rating']=data.groupby('Category' )['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
data.isnull().sum()
```

```python
# Standardized column names using snake_case for consistency and easier reference during analysis.
```

```python
data.columns=data.columns.str.lower()
data.columns=data.columns.str.replace(' ','_')
```

```python
data=data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
data.columns
```

```python
# create a new column age_group
```

```python
labels=['Young Adult','Adult','Middle_aged','Senior']
data['age_group']=pd.qcut(data['age'],q=4,labels=labels)
```

```python
data[['age','age_group']].head()
```

```python
# create new column purchase_frequency_days
```

```python

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

data['purchase_frequency_days'] = data['frequency_of_purchases'].map(frequency_mapping)
```

```python
data[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
# Dropping promo code used column

data = data.drop('promo_code_used', axis=1)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

params = quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=SMILE;"
    "DATABASE=customer_behaviour;"
    "Trusted_Connection=yes;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={params}"
)

print("Engine Created Successfully!")
```

```python
data.to_sql(
    name='customer_shopping_behavior',
    con=engine,
    if_exists='replace',
    index=False
)

print("Table Created Successfully!")
```
